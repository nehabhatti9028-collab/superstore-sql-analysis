```python
import pandas as pd
import sqlite3

df = pd.read_csv("/content/Sample - Superstore.csv", encoding="latin-1")
conn = sqlite3.connect("superstore.db")
df.to_sql("orders", conn, if_exists="replace", index=False)

pd.read_sql("SELECT Category, SUM(Sales) AS total_sales, SUM(Profit) AS total_profit FROM orders GROUP BY Category", conn)
```

```python
pd.read_sql("""
SELECT "Customer Name", ROUND(SUM(Sales), 2) AS total_spent
FROM orders
GROUP BY "Customer Name"
ORDER BY total_spent DESC
LIMIT 10
""", conn)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Year"] = df["Order Date"].dt.year
df.to_sql("orders", conn, if_exists="replace", index=False)

pd.read_sql("""
SELECT Year, ROUND(SUM(Sales), 2) AS total_sales, ROUND(SUM(Profit), 2) AS total_profit
FROM orders
GROUP BY Year
ORDER BY Year
""", conn)
```

```python
pd.read_sql("""
SELECT "Sub-Category", ROUND(SUM(Sales), 2) AS total_sales, ROUND(SUM(Profit), 2) AS total_profit
FROM orders
GROUP BY "Sub-Category"
HAVING SUM(Profit) < 0
ORDER BY total_profit
""", conn)
```

```python
pd.read_sql("""
SELECT "Product Name", ROUND(SUM(Profit), 2) AS total_profit
FROM orders
GROUP BY "Product Name"
ORDER BY total_profit DESC
LIMIT 5
""", conn)
```

## Key Insights
1. Sales grew about 1.5x from 2014 to 2017 and profit about 1.9x.
2. Profit margin rose from about 10% (2014) to about 12.7% (2017).
3. Only 3 sub-categories lose money: Tables, Bookcases and Supplies.
4. Tables have the biggest loss (-17.7K) despite over 2 lakh in sales.
5. Top customer spent over 25K. The top profit product is a Canon copier (about 25K profit).
